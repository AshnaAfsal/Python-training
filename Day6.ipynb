class Solution:
    def floodFill(self, image: List[List[int]], sr: int, sc: int, color: int) -> List[List[int]]:
        org = image[sr][sc]      # not image[sc][sr]

        if org == color:
            return image

        def Fill(i, j):
            if (i < 0 or i >= len(image) or
                j < 0 or j >= len(image[0]) or
                image[i][j] != org):
                return

            image[i][j] = color   # not image[sr][sc]

            Fill(i - 1, j)
            Fill(i + 1, j)
            Fill(i, j - 1)
            Fill(i, j + 1)

        Fill(sr, sc)             # not Fill(i, j)
        return image

class Solution:
    def numIslands(self, grid: List[List[str]]) -> int:
        if not grid:
            return 0
        def count_island(grid,i,j):
            if i<0 or i>=len(grid) or j<0 or j>=len(grid[0]) or grid[i][j]!='1':
                return
            grid[i][j]='2'
            count_island(grid,i+1,j)
            count_island(grid,i-1,j)
            count_island(grid,i,j+1)
            count_island(grid,i,j-1)

        c=0
        for i in range(len(grid)):
            for j in range(len(grid[0])):
                if grid[i][j]=='1':
                    count_island(grid,i,j)
                    c+=1
        return c
      ```
