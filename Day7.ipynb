class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        def poss(board,i,j,k):
            if k==len(word):
                return True
            if i<0 or i>=len(board) or j<0 or j>=len(board[0]) or board[i][j]!=word[k]:
                return False
            temp=board[i][j]
            board[i][j]="#"
            if poss(board,i-1,j,k+1) or poss(board,i+1,j,k+1) or poss(board,i,j-1,k+1) or poss(board,i,j+1,k+1):
                return True
            board[i][j]=temp
            return False
        for i in range(len(board)):
            for j in range(len(board[0])):
                if poss(board,i,j,0):
                    return True
        return False

class Solution:
    def isValid(self, s: str) -> bool:
        l = []

        for ch in s:
            if ch == '(':
                l.append(')')
            elif ch == '[':
                l.append(']')
            elif ch == '{':
                l.append('}')
            else:
                if not l or l.pop() != ch:
                    return False

        return len(l) == 0

class Solution:
    def backspaceCompare(self, s: str, t: str) -> bool:
        
        def check(s):
            l=[]
            for c in s:
                if c!='#':
                    l.append(c)
                    
                else:
                    if l:
                        l.pop()
            return "".join(l)
        x=check(s)
        y=check(t)
        
        return x==y

class Solution:
    def countStudents(self, students: List[int], sandwiches: List[int]) -> int:
        n=len(students)
        while students and sandwiches and sandwiches[0] in students:
            if students[0]!=sandwiches[0]:
                students.append(students.pop(0))
            else:
                n-=1
                students.pop(0)
                sandwiches.pop(0)
        return n

lass Solution:
    def nextGreaterElement(self, nums1: List[int], nums2: List[int]) -> List[int]:  
        d={}
        l=[]
        for i in nums2:
            while l and l[-1]<i:
                s=l.pop()
                d[s]=i
            l.append(i)
        while l:
            d[l.pop()]=-1
        r=[]
        for j in nums1:
            r.append(d[j])

class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        l=[]
        n=len(temperatures)
        r=[0]*n
        for i in range(n-1,-1,-1):
            while l and temperatures[i]>=temperatures[l[-1]]:
                l.pop()
            if l:
                r[i]=l[-1]-i
            l.append(i)
        return r
