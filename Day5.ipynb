`````python
#given a list and k value we want to find the sum of the list which has maximum length
li=[2,1,6,2,1,2,4,7,8,1,2]
k=10
r,l=0,0
s=0
m=0
while r<len(li):
  s+=li[r]
  while s>k:
    s-=li[l];
    l+=1
  length=r-l+1
  m=max(m,length)
  r+=1
print(m)

#dynamic sliding
`````
