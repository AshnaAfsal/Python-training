```python
l=list(map(str,input().split()))
res=[]
for i in l:
  if i not in res:
    res.append(i)
print(res)
```

```python
l=list(map(str,input().split()))
res=[]
for i in l:
  if l.count(i)%2!=0 and i not in res:
    res.append(i)
print(res)
```

```python
l=list(map(int,input().split()))
l.sort()
r=[]
for i in l:
  if i%2!=0:
    r.append(i)
  else:
    r.insert(0,i)
print(r)
```
