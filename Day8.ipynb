        slow=head
        while fast and fast.next:
            fast=fast.next.next
            slow=slow.next
            if fast==slow:
                fast=head
                while fast!=slow:
                    fast=fast.next

    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        fast=head
        slow=head
        for i in range(n):
            fast=fast.next
        if fast is None:
            return head.next

        while fast and fast.next:
            fast=fast.next
            slow=slow.next
        slow.next=slow.next.next

            
        return head

            prev=curr
