````python
#k rotations left rotation brute force approach
l=[1,2,3,4,5]
k=int(input("Enter the number of rotations"))

for i in range(k):
  temp=l[0]
  for i in range(1,len(l)):
    l[i-1]=l[i]
  l[-1]=temp
print(l)
````

````python

#k rotations left rotation
l=[1,2,3,4,5,6,7]
k=int(input("Enter the number of rotations"))
n=len(l)
k=k%n
def rotate(i,j):
  while i<j:
    l[i],l[j]=l[j],l[i]
    i+=1
    j-=1
rotate(0,k-1)
print(l)
rotate(k,n-1)
print(l)
rotate(0,n-1)
print(l)
````

````python
#k rotation right l=[1,2,3,4,5,6,7]
k=int(input("Enter the number of rotations"))
n=len(l)
k=k%n
def rotate(i,j):
  while i<j:
    l[i],l[j]=l[j],l[i]
    i+=1
    j-=1
# Steps for right rotation using reversal algorithm:
# 1. Reverse the first (n-k) elements
# 2. Reverse the last k elements
# 3. Reverse the entire list

rotate(0, n-k-1)
print(l)
rotate(n-k, n-1)
print(l)
rotate(0, n-1)
print(l)
````
