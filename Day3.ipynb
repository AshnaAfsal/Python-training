        l=len(nums)

class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        if n<1:
            return False
        while n:
            if n==1:
                return True
            elif n%2!=0:
                return False
            else:
                n=n//2

````python
#left rotating the list
l=[1,2,3,4,5]
n=len(l)
temp=l[0]
for i in range(0,n-1):
  l[i]=l[i+1]
l[n-1]=temp
print(l)
````
