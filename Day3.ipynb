```
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        for i in range(len(digits)-1, -1,-1):
            if  digits[i]==9:
                digits[i]=0
            else:
                digits[i]=digits[i]+1
                return digits
        return [1]+digits
#leetcode 66
```

```
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        x=0
        for i in nums:
            x=x^i
        return x
#leetcode 136
```

```
class Solution:
    def missingNumber(self, nums: List[int]) -> int:
        l=len(nums)        
        x,y=0,0
        for i in range (l+1):
            x=x^i
        for i in nums:
            y=y^i
        return x^y
#leetcode 268
```

```
class Solution:
    def missingNumber(self, nums: List[int]) -> int:
        l=len(nums)
        s=(l*(l+1))//2
        s1=sum(nums)
        n=s-s1
        return n
#leetcode 268
```

        if n<1:
            return False
        while n:
            if n==1:
                return True
            elif n%2!=0:
                return False

```
class Solution:
    def isPowerOfTwo(self, n: int) -> bool:
        if n<1:
            return False
        if n&(n-1)==0:
            return True
        else:
            return False

#231
```
